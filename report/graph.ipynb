```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
openMP = pd.read_excel("results.xlsx", sheet_name="OpenMP0")
omp_means = [openMP[col].mean() for col in openMP]
omp_speedups = [ omp_means[0]/x for x in omp_means]

# labels = openMP.columns
labels = [1, 2, 4, 8, 12]
print(omp_speedups)
```

```python
MPI = pd.read_excel("results.xlsx", sheet_name="MPI")
mpi_means = [MPI[col].mean() for col in MPI]
mpi_speedups = [ mpi_means[0]/x for x in mpi_means]
print(mpi_speedups)
```

```python
fig, (ax, ax1) = plt.subplots(1, 2, figsize=(10, 4))
ax.plot(labels, omp_speedups, 'x-', label="OpenMP")
ax.plot(labels, mpi_speedups, 'x-', label="MPI")
ideal_speedups = [1, 2, 4, 8, 12]
ax.plot(labels, ideal_speedups, 'x-', label="Идеальное ускорение")
ax.legend()

ax1.plot(labels, omp_speedups, 'x-', label="OpenMP")
ax1.plot(labels, mpi_speedups, 'x-', label="MPI")
ax1.legend()

fig.supxlabel("Количество потоков")
fig.supylabel("Ускорение, раз")
fig.suptitle("Ускорение в зависимости от количества потоков, раз")
# plt.savefig("results.pdf", format="pdf")

plt.show(ax)
```
